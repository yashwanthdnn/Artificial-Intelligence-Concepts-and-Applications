```python
responses = {
    "hi": "Hello! How can I help you?",
    "hello": "Hi there! What can I do for you?",
    "hey": "Hey! How can I assist you today?",
    "good morning": "Good morning! How can I help?",
    "good afternoon": "Good afternoon! What can I do for you?",
    "good evening": "Good evening! How can I help you?",
    "how are you": "I am just a program, but I am doing fine!",
    "what is your name": "I am a simple chatbot.",
    "your name": "I am a simple chatbot.",
    "who created you": "I was created as a simple Python chatbot example.",
    "what can you do": "I can respond to greetings, answer a few basic questions, and say goodbye.",
    "help": "You can say hi, ask my name, ask how I am, or type bye to exit.",
    "thanks": "You're welcome!",
    "thank you": "You're welcome!",
    "ok": "Alright! Ask me anything else.",
}

print("Simple Chatbot")
print("Type 'bye' to exit.\n")

while True:
    user_input = input("You: ").strip().lower()

    if user_input in ("bye", "exit", "quit"):
        print("Bot: Goodbye!")
        break

    reply = responses.get(user_input, "Chat Bot: Sorry, I do not understand that. Try 'help'.")
    print(f"Bot: {reply}")
```
