```python
df["Date"]=pd.to_datetime(df["Date"])
df.info()
```

```python
df["Day"] = df["Date"].dt.day
df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year

df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated
```

```python
from sklearn.model_selection import train_test_split

x = df[['Day', 'Month', 'Year']]
y = df['Sales']

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
predections=model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae=mean_absolute_error(y_test,predections)
print("The Absolute Error:",round(mae,2))
```

```python
future=pd.DataFrame({
    'Day':[11,12,13],
    'Month':[1,1,1],
    'Year':[2024,2024,2024]
})
```

```python
future_prediction=model.predict(future)
print("\nFuture demand Prediction")
for day,demand in zip(future['Day'],future_prediction):
    print(f"Day {day}:{round(demand)}units")
```
