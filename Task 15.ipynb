```python
questions = [

  "Where is my order?",

  "Can I track my package?",

  "What is the status of my order?",

  "When will my order arrive?",

  "I have not received my order",



  "I want a refund",

  "How can I get my money back?",

  "I want to return my product",

  "Can I get a refund for my purchase?",

  "I need to cancel and get a refund",



  "I forgot my password",

  "How do I reset my password?",

  "I cannot login to my account",

  "I lost my account password",

  "Help me change my password",



  "My payment failed",

  "Why was my payment declined?",

  "I cannot make a payment",

  "My card payment is not working",

  "There is a problem with my payment",



  "I want to know about your products",

  "Do you have this product?",

  "Is this product available?",

  "Tell me about the product",

  "What products do you sell?",



  "What are your working hours?",

  "When is customer support available?",

  "Are you open today?",

  "What time does support close?",

  "When can I contact customer service?"

]



labels = [

  "order", "order", "order", "order", "order",

  "refund", "refund", "refund", "refund", "refund",

  "password", "password", "password", "password", "password",

  "payment", "payment", "payment", "payment", "payment",

  "product", "product", "product", "product", "product",

  "hours", "hours", "hours", "hours", "hours"

]
```

```python
import re

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

cleaned_questions = [preprocess(q) for q in questions]
```

```python
cleaned_questions
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(cleaned_questions)
```

```python
#see all tokens/features
tokens = vectorizer.get_feature_names_out()
print(tokens)
```

```python
for questions in cleaned_questions:
    tokens=vectorizer.build_analyzer()(questions)
    print(tokens)
```

```python
print(X)
```

```python
#logistic regression model
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X, labels)



```

```python
responses={
    "order":"You can track your order using the tracking number provided in your confirmation email. If you haven't received it, please contact our support team.",
    "refund":"To request a refund, please visit our returns page and follow the instructions. Make sure to have your order number ready.",
    "password":"If you've forgotten your password, you can reset it by clicking on the 'Forgot Password' link on the login page. Follow the instructions to create a new password.",
    "payment":"If your payment failed, please check your card details and ensure you have sufficient funds.",
    "product":"We offer a wide range of products. You can browse our catalog on our website or contact our support team for more information.",
    "hours":"Our customer support is available from 9 AM to 5 PM, Monday to Friday. You can reach us via phone, email, or live chat during these hours."
}
```

```python
#Customer Support Chatbot

print("Customer Support Chatbot")
print("Type 'exit' to end the conversation.")
while True:
    user_input = input("You: ")
    if user_input.lower() == 'exit':
        print("Chatbot: Thank you for using our customer support chatbot. Goodbye!")
        break

    cleaned_input = preprocess(user_input)
    input_vector = vectorizer.transform([cleaned_input])
    prediction = model.predict(input_vector)[0]

    response = responses.get(prediction, "I'm sorry, I didn't understand that. Can you please rephrase?")
    print(f"Chatbot: {response}")
```
